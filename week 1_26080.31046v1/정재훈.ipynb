## 1-1 RL이란?
강화학습
시행착오를 거치며 가장 높은 보상(Reward)을 얻는 행동 법칙을 스스로 찾아내는 학습법

- State ($s_t$): 현재까지 생성된 프롬프트 및 프리픽스 텍스트 ($y_{<i}$)
- Action ($a_t$): 어휘집(Vocabulary)에서 선택하여 출력할 다음 토큰 ($y_i$)
- Policy(Agent 내부 부분) ($\pi_\theta$): 현재 파라미터 $\theta$ 기반의 다음 토큰 확률 분포 $P(y_i \mid y_{<i})$
- Reward ($r_t$): 생성된 토큰 또는 전체 궤적에 대해 부여되는 보상 신호

## 1-2. Policy Gradient 수식 이해

핵심 수식: $\nabla_\theta \mathcal{L} = - A_i \cdot \nabla_\theta \log \pi_\theta(y_i \mid x, y_{<i})$

본문:어드밴티지 ($A_i$): 현재 행동이 기대치보다 얼마나 더 좋은가/나쁜가를 나타내는 수치
- $A_i > 0$ (양수 보상): "잘했다" $\rightarrow$ 해당 토큰의 생성 확률 $\pi_\theta(y_i)$을 높임
- $A_i < 0$ (음수 보상): "못했다" $\rightarrow$ 해당 토큰의 생성 확률 $\pi_\theta(y_i)$을 억제(Suppression)함

---
각 요소 설명
- $\theta$ (파라미터): 언어 모델(LLM) 내부의 가중치(Weights) 집합입니다.
- $\mathcal{L}$ (손실 함수, Loss Function): 모델을 경사하강법(Gradient Descent)으로 업데이트하기 위해 정의한 손실 값입니다.
- $\pi_\theta(y_i \mid x, y_{<i})$ (정책 / 단어 생성 확률):질문 $x$와 지금까지 생성된 이전 토큰들 $y_{<i}$가 주어졌을 때, 모델이 $i$번째 토큰으로 $y_i$를 선택할 확률입니다.
- $\log \pi_\theta(\dots)$ (로그 확률):확률값(0~1 사이)에 로그를 취한 것입니다. 수치적 안정성과 미분의 편의성을 위해 사용됩니다.
- $\nabla_\theta \log \pi_\theta(\dots)$ (로그 확률의 기울기):"이 토큰($y_i$)을 선택할 확률을 높이려면 모델 파라미터 $\theta$를 어느 방향으로 이동시켜야 하는가?"를 나타내는 변화의 방향 벡터입니다.
- $A_i$ (어드밴티지, Advantage):$i$번째 토큰 생성 행동이 "평균적인 행동보다 얼마나 더 좋은가?"를 나타내는 스칼라 점수입니다.양수($A_i > 0$)이면 훌륭한 선택, 음수($A_i < 0$)이면 엉망인 선택이었음을 뜻합니다.

## 2.1 토큰 단위의 Dense Reward의 필요성

### 1. 기술적 핵심 개념 (왜 이 흐름이 중요한가?)

#### 🔴 기존 방식: Outcome-based Sparse Reward (예: RLVR, GRPO)

* **작동 방식:** 모델이 전체 생각의 사슬(Chain-of-Thought, CoT)과 최종 정답을 모두 생성한 후, 맨 마지막에 오직 하나의 보상(예: $+1$ 또는 $0$)만 부여합니다.
* **치명적인 한계 (Credit Assignment Problem):**
* 예를 들어 1,000개 토큰으로 구성된 긴 풀이 과정에서 950번째 토큰 하나 때문에 틀렸다고 가정해 보겠습니다.
* Sparse Reward 방식에서는 전체 결과가 오답($0$점)이므로, 앞선 949개의 **훌륭한 추론 토큰들까지 전부 음수/불이익(Penalty) 방향으로 업데이트**됩니다.
* 즉, "어느 부분에서 잘했고 어디서 틀렸는지" 신호를 세밀하게 전달하지 못해 **학습 효율이 극도로 저하**되고 **High Variance(높은 분산)** 문제가 발생합니다.



#### 🟢 개선 방식: Token-level Dense Signal (예: OPD)

* **작동 방식:** 교사(Teacher) 모델의 확률 분포나 중간 상태를 활용하여, 모델이 생성하는 **매 토큰($t=1, 2, \dots, T$) 단위마다 Advantage(상대적 우수성) 신호**를 전달합니다.
* **해결책 (Dense Feedback):**
* 각 단계마다 촘촘한(Dense) 지도 신호를 주므로, 중간에 발생한 작은 오류를 즉각 교정하고 올바른 추론 경로로 유도할 수 있습니다.
* 결과적으로 학습이 훨씬 안정적이고 속도가 빠르며, 긴 추론 과정(Long-horizon Reasoning)을 효과적으로 학습할 수 있게 됩니다.


## 2.2 OPD가 무엇인가

### 1. **핵심 배경: 기존 증류(Offline Distillation / Supervised KD)의 한계**

* **기존 방식:** 교사 모델(Teacher)이 미리 생성해 둔 정답 문장이나 고정된 데이터셋(Ground-truth)을 바탕으로 학생 모델(Student)을 학습시켰습니다.
* **문장의 누적 오류 문제 (Exposure Bias / Distribution Mismatch):**
* 문장을 생성하는 LLM(Autoregressive Model)은 이전까지 자기가 쓴 단어를 바탕으로 다음 단어를 예측합니다.
* 기존 방식처럼 교사의 깔끔한 문장만 보며 학습한 학생은, 실제 추론(Inference) 과정에서 **자신이 중간에 단어를 잘못 선택했을 때** 이를 어떻게 수습해야 할지 알지 못합니다.
* 그 결과 한 번 오답 단어가 나오면 오류가 눈덩이처럼 불어나며 전혀 이상한 문장을 생성(Hallucination / Degeneration)하게 됩니다.



---

### 2. **Slide 7의 핵심 개념: On-Policy Rollout & 토큰 단위 피드백**

Slide 7에서 말하는 On-Policy Distillation (OPD)은 강회학습(RL) 및 모의 학습(Imitation Learning)의 핵심 개념을 언어 모델 증류에 도입한 것입니다.

* **On-Policy Rollout (학생의 직접 문장 작성):**
* 교사가 작성해 준 답안지를 따라 읽게 하는 것이 아니라, 학생 모델이 현재 시점의 파라미터 상태로 직접 문장을 끝까지 생성(Sampling)해 보게 합니다.
* 이 과정에서 학생은 자신이 잘한 부분뿐만 아니라 **실수로 틀린 문맥/단어**까지 포함된 문장을 스스로 만들어냅니다.


* **토큰 단위 채점 및 피드백 (Teacher Feedback):**
* 학생이 삐끗해서 작성 중인 문장($y_{<i}$)을 문맥 조건으로 주었을 때, "교사 모델이라면 이 문맥 뒤에 어떤 단어를 썼을 것인가?"를 실시간으로 질문합니다.
* 교사 모델은 학생이 오답을 내놓은 시점마다 토큰 단위의 확률 분포(Logits)를 계산하여, 학생이 작성한 각 토큰이 얼마나 타당한지 채점해 줍니다.

### 3. **비유를 통한 쉬운 이해**

* **기존 방식 (Supervised KD):**
> 선생님이 풀어놓은 **100점짜리 풀이 과정**만 집에서 백날 복습하는 방식입니다. 시험장에 가서 자기가 계산 실수를 하나라도 하면, 그 이후 과정을 어떻게 풀어나가야 할지 몰라 시험을 망치게 됩니다.


* **On-Policy Distillation (OPD):**
> 학생이 직접 **모의고사 답안**을 작성합니다. 도중에 오답이나 이상한 길로 빠지더라도, 옆에 서 있는 선생님이 **"네가 지금 $i$번째 단계에서 이 단어를 택했는데, 그 상황에서는 그 단어보다 이 단어를 쓸 확률이 더 높단다"** 하고 매 단계(Token)마다 실시간 피드백을 주는 방식입니다.



---

### 4. **OPD가 가져오는 이점 및 효과**

1. **실수로부터의 학습 (Learning from Self-Generated Mistakes):**
* 학생 모델은 자기가 자주 범하는 실수 패턴과 잘못된 문맥 상황을 경험하고, 그 상태를 어떻게 정상으로 되돌리는지 교사의 확률 분포를 통해 배웁니다.


2. **Train-Inference Distribution Mismatch 해결:**
* 학습할 때 겪는 문장 상태 분포와 실제 서비스 추론 시 겪는 문장 상태 분포가 일치하게 되므로, 추론 시 문장 생성 성능 및 안전성이 크게 향상됩니다.


3. **수식과의 연결 (Slide 8 예고):**
* 학생이 선택한 토큰 $y_i$에 대해 교사의 확률 $\pi_t(y_i)$가 학생의 확률 $\pi_s(y_i)$보다 높으면 칭찬(양의 어드밴티지)을, 반대인 경우 벌점(음의 어드밴티지)을 주어 학생이 교사의 선호 분포를 정확히 따라가도록 유도합니다.

## 2-3. OPD의 수학적 정의 (Reverse KL Divergence)


### 1. **목적 함수 (Objective Function): Reverse KL Divergence**

On-Policy Distillation (OPD)의 핵심은 **학생 모델이 스스로 생성한 문맥(On-policy distribution) 위에서** 교사 모델과의 차이를 줄이는 것입니다. 이를 수학적으로 나타내기 위해 **Reverse KL Divergence**를 손실 함수(Loss Function)로 사용합니다.

$$\mathcal{L}_{\text{OPD}}(\theta) = \mathbb{E}_{y \sim \pi_{\theta}(\cdot\vert{}x)} \left[ D_{\text{KL}}(\pi_{\theta}(\cdot\vert{}x, y_{<i}) \parallel \pi_t(\cdot\vert{}x, y_{<i})) \right]$$

* **$\pi_{\theta}$**: 파라미터 $\theta$를 가진 **학생 모델 (Student Model)**
* **$\pi_t$**: 고정된 **교사 모델 (Teacher Model)**
* **$x$**: 입력 프롬프트 / 프롬프트 조건
* **$y_{<i}$**: 학생 모델이 $i-1$번째 토큰까지 스스로 생성해 온 이전 문맥 (On-Policy Rollout)

---

### 2. **수식 분해 및 그래디언트(Gradient) 도출**

손실 함수를 최소화하기 위해 $\theta$에 대해 그래디언트를 취하면, 강회학습의 **Policy Gradient (REINFORCE)** 형태 또는 **가중된 Cross-Entropy** 형태로 전개됩니다.

#### 토큰 단위 Log Likelihood 표현식:

$$\mathcal{L}_{\text{OPD}}(\theta) = \mathbb{E}_{y \sim \pi_{\theta}} \left[ \sum_{i=1}^{T} \sum_{v \in V} \pi_{\theta}(v \vert{} x, y_{<i}) \log \frac{\pi_{\theta}(v \vert{} x, y_{<i})}{\pi_t(v \vert{} x, y_{<i})} \right]$$

이를 그래디언트 형태(Policy Gradient Theorem 적용)로 나타내면 다음과 같습니다:

$$\nabla_{\theta} \mathcal{L}_{\text{OPD}}(\theta) = \mathbb{E}_{y \sim \pi_{\theta}} \left[ \sum_{i=1}^{T} \underbrace{\nabla_{\theta} \log \pi_{\theta}(y_i \vert{} x, y_{<i})}_{\text{학생의 확률 조절}} \times \underbrace{A(x, y_{\le i})}_{\text{어드밴티지 (Reward/Penalty)}} \right]$$

---

### 3. **어드밴티지 $A(x, y_{\le i})$의 의미 해석**

수식에서 각 토큰 $y_i$를 생성했을 때 적용되는 보상/벌점(어드밴티지 $A$)은 다음과 같이 계산됩니다:

$$A(x, y_{\le i}) = \log \pi_t(y_i \vert{} x, y_{<i}) - \log \pi_{\theta}(y_i \vert{} x, y_{<i})$$

또는 교사와 학생의 토큰 분포 전체 간의 Log Odds 차이로 정의됩니다:

* **$A > 0$ (교사 확률 > 학생 확률):**
학생이 선택한 토큰 $y_i$에 대해 교사의 확률이 더 높다면, 학생의 행동은 **좋은 선택**으로 평가받고 해당 토큰의 생성 확률을 **상승(Encourage)** 시킵니다.
* **$A < 0$ (교사 확률 < 학생 확률):**
학생이 선택한 토큰 $y_i$에 대해 교사의 확률이 더 낮다면, 학생의 행동은 **잘못된 선택**으로 평가받고 해당 토큰의 생성 확률을 **하락(Suppress)** 시킵니다.

---

### 4. **Forward KL vs. Reverse KL 비교 (왜 Reverse KL인가?)**

| 구분 | Forward KL ($D_{\text{KL}}(\pi_t \parallel \pi_{\theta})$) | Reverse KL ($D_{\text{KL}}(\pi_{\theta} \parallel \pi_t)$) |
| --- | --- | --- |
| **기초 접근법** | 기존 Supervised KD (Offline) | **OPD (On-Policy)** |
| **샘플링 주체** | 교사 분포 또는 Ground-truth 데이터셋 ($y \sim \pi_t$) | **학생 직접 생성 데이터 ($y \sim \pi_{\theta}$)** |
| **특징** | **Zero-avoiding:** 교사가 확률을 부여한 모든 곳을 학생이 가버려야 하므로 불확실성이 커짐 (Mode Averaging). | **Zero-forcing:** 학생이 생성한 경로 위에서 교사의 모드(Mode)에 선명하게 집중함 (Mode Covering). |
| **결과** | 추론 시 오류 누적(Exposure Bias) 발생 가능성이 높음. | 학생이 생성하는 문장 범위 내에서 **매우 선명하고 정확한 오답 피드백** 가능. |

## 3-1. 본 논문의 핵심 질문 (Core Question)

**"교사 모델은 과연 학생이 작성한 오프폴리시(Off-policy) 문맥에서도 신뢰할 수 있는 지도 신호를 제공하는가?"**

## 3-2. 충격적 발견 1 - 교사의 피드백은 극심한 노이즈 투성이다


### 1. 배경: 온폴리시 증류(OPD)와 교사 모델의 역할

온폴리시 증류(OPD)는 학생 모델(Student)이 생성한 토큰들에 대해 교사 모델(Teacher)이 토큰 단위 평가 신호인 어드밴티지(Advantage, $A_i$)를 계산하여 피드백을 주는 방식입니다.

* **이상적인 상태:** 학생이 작성한 코딩/수학 풀이에서 정답으로 유도하는 좋은 토큰에는 칭찬($A_i > 0$), 틀린 길로 빠지는 오답 토큰에는 감점($A_i < 0$)을 부여해야 합니다.

---

### 2. 노이즈(Noise)의 정의

논문에서는 교사 모델의 피드백 신호와 실제 최종 정답의 정확도(Ground Truth Correctness)가 충돌하는 경우를 **노이즈**로 정의합니다.

1. **False Negative (억울한 감점):** 학생이 최종적으로 **정답**을 맞힌 트래젝터리(풀이 과정)임에도 불구하고, 교사가 해당 토큰에 음수의 어드밴티지($A_i < 0$)를 주는 경우
2. **False Positive (잘못된 칭찬):** 학생이 최종적으로 **오답**을 냈음에도 불구하고, 교사가 그 오답으로 가는 토큰들에 양수의 어드밴티지($A_i > 0$)를 주는 경우

---

### 3. 왜 교사 모델이 커질수록 노이즈가 더 심해질까?

실험 결과(Qwen3 계열, 학생 모델 1.7B 기준):

* **4B 교사:** 30.6% 노이즈
* **30B 교사:** 34.7% 노이즈
* **235B 교사:** **50.6% 노이즈** (동전 던지기 수준)


### 4. 정답을 맞춰도 감점때림!
1. \boxed{} 토큰이 왜 중요한가?수학/추론 문제 해설 데이터셋(MATH, AIME 등)에서 \boxed{정답}은 최종 정답이 제출되는 위치를 나타내는 가장 결정적인 지점입니다.즉, 학생 모델이 문제를 쭉 풀어나간 뒤 \boxed{15}라고 썼고 그 15가 실제로 Ground Truth(검증된 정답)라면, \boxed{} 토큰 및 그 직후 정답 토큰은 논리적으로 무조건 칭찬(양수 어드밴티지, $A_i > 0$)을 받아야 하는 지점입니다.
2. Qwen3-235B 교사의 충격적인 행동실험에서 1.7B 학생 모델이 생성한 트래젝터리를 분석해 본 결과:교사 모델의 체급이 커질수록 정답 위치에서의 억울한 벌점(False Negative) 부여율이 폭증했습니다.특히 Qwen3-235B 초대형 교사의 경우, 학생이 최종 정답을 정확히 맞혔음에도 불구하고 \boxed{} 토큰의 97.8%에 대해 음수의 어드밴티지($A_i < 0$, 벌점)를 부여했습니다.사실상 "학생이 정답을 맞힌 모든 상황에서 벌점을 때렸다"고 해도 무방한 수준입니다.

#### 원인: 분포 불일치(Distributional Mismatch) 현상

* **표현 방식의 차이:** 초대형 교사 모델(235B)은 소형 학생 모델(1.7B)과 문맥을 이해하고 문장을 생성하는 스타일/확률 분포의 차이가 매우 큽니다.
* **과도한 감점(Over-pessimism):** 교사 모델의 규모가 커질수록 "자신이 선호하는 표현 방식"과 조금만 달라도 학생 모델이 생성한 모든 토큰에 **음수의 어드밴티지($A_i < 0$)를 남발**하는 경향이 강해집니다.
* 그 결과, 학생 모델이 정답을 맞혔음에도 교사 모델은 학생의 생소한 풀이 스타일을 틀린 것으로 간주하여 감점을 주는 비율이 폭증하게 됩니다.



## 3-5. 충격적 발견 2 - 노이즈 피드백만 모아 학습시켜도 똑같이 성장한다



### 1. 실험 설계 (Ablation Study)

연구진은 앞서 밝혀낸 "교사 피드백의 극심한 노이즈"가 실제로 학습에 어떤 영향을 미치는지 검증하기 위해 피드백 신호를 인위적으로 분리하여 비교하는 실험을 진행했습니다.

* **Standard OPD:** 기존 방식대로 교사 모델이 준 전체 피드백(노이즈+정제 신호 혼합)으로 학습
* **Clean-only:** 정답에는 칭찬, 오답에는 감점을 정확히 준 **100% 올바른 피드백**만 선별하여 학습
* **Noisy-only:** 정답에 감점, 오답에 칭찬을 준 **100% 엉터리/노이즈 피드백**만 골라내어 학습

---

### 2. 실험 결과 및 의미

* **결과:** 세 그룹의 **학습 수렴 곡선(Convergence Curve)과 최종 추론 Performance(정확도)가 완전히 일치**했습니다.
* **의미:** 교사 모델이 아무리 엉터리 신호(Noisy)를 주든, 완벽한 신호(Clean)를 주든 학생 모델의 성장에 영향이 없었습니다.
* 이는 OPD에서 학생 모델이 성능이 오르는 이유가 **"교사가 정답/오답을 잘 가르쳐줘서(Distillation)"가 아니라는 것**을 완벽하게 입증합니다.

---

### 3. 왜 이런 기현상이 발생하는가? (학습 메커니즘)

#### ① 확률 억제 메커니즘 (Probability Suppression)

학생 모델 입장에서 교사의 어드밴티지 신호 $A_i$가 음수($A_i < 0$, 감점)이면, 해당 토큰을 등장시킬 확률을 낮추는 방향으로 손실 함수(Loss)가 작동합니다.

* 교사는 분포 불일치 때문에 학생의 거의 모든 토큰(특히 235B의 경우 정답을 포함한 97.8%)에 **음수의 어드밴티지**를 남발합니다.
* 역설적이게도 이 "무차별 감점" 신호가 학생 모델 입장에서 **자신이 생성한 궤적(Trajectory) 전체의 비중을 낮추고, 불확실성이 높은 오답 탐색을 억제**하는 효과를 냅니다.

#### ② '증류(Distillation)'가 아니라 '자기 개선(Self-Improvement)'이다

* 교사 모델은 고품질의 지식을 전달하는 '스승'이 아니라, 단지 학생이 스스로 다양한 답안을 생성해 보고 그중 불필요한 탐색을 줄여나가도록 돕는 **'랜덤성 억제 장치(Squelch)'** 역할만 수행하고 있었던 것입니다.
* 결과적으로 학생 모델은 교사의 지식을 흡수해서 똑똑해진 것이 아니라, 온폴리시(On-Policy) 환경에서 자신의 토큰 생성을 정제해 나가는 **Self-Improvement(자기 개선)** 과정으로 성장을 이뤄냈습니다.


## 3-6. 성능 향상의 진짜 원인 탐색 1 - 어디서 학습이 일어나는가?


### 1. 그래디언트 수식의 의미

온폴리시 증류나 정책 단계를 업데이트할 때, 손실 함수에 들어가는 핵심 그래디언트 항은 다음과 같습니다.

$$\nabla_\theta \log \pi_\theta(y_i) = \frac{\nabla_\theta \pi_\theta(y_i)}{\pi_\theta(y_i)}$$

* $\pi_\theta(y_i)$: 학생 모델이 현재 문맥에서 $y_i$라는 토큰을 출력할 **확률값 ($0 < \pi \le 1$)**
* $\nabla_\theta \pi_\theta(y_i)$: 모델 매개변수 $\theta$에 대한 확률의 미분값
* **핵심 포인트:** 로그 미분 공식($\frac{d}{dx}\ln x = \frac{1}{x}$)에 의해, 확률 $\pi_\theta(y_i)$가 **분모**로 내려가게 됩니다.

---

### 2. Head 토큰 vs Tail 토큰 동작 원리

학생 모델이 생성하는 토큰들은 모델이 부여하는 확률값($\pi$)의 크기에 따라 두 가지로 나뉩니다.

#### ① Head 토큰 (상위 확률 토큰, 예: 상위 80%)

* **개념:** 모델이 이미 완벽히 학습하여 "매우 높은 확신"을 갖고 출력하는 토큰들입니다 (예: 문법적 조사, 당연한 다음 단어).
* **수학적 현상:**
1. 소프트맥스(Softmax) 함수의 특성상, 특정 토큰의 확률 $\pi_\theta(y_i)$가 $1$에 가까워지면 해당 토큰의 확률 변화율인 $\nabla_\theta \pi_\theta(y_i)$는 $0$으로 수렴합니다 (위로 솟은 곡선의 꼭짓점 부근).
2. 따라서 분모가 크더라도 분자 자체가 $0$에 가까워져 최종 그래디언트 $\nabla_\theta \log \pi_\theta(y_i) \approx 0$이 됩니다.


* **결과:** 교사가 무슨 피드백(노이즈/벌점 등)을 주든, Head 토큰 영역에서는 **학습(파라미터 업데이트)이 거의 일어나지 않습니다.**

#### ② Tail 토큰 (하위 확률 토큰, 예: 하위 20%)

* **개념:** 모델이 확신이 없어 "불안정하거나 낮은 확률"로 얼떨결에 생성한 토큰들입니다 (예: 복잡한 추론 단계, 어색한 단어 선택).
* **수학적 현상:**
1. 확률 $\pi_\theta(y_i)$ 값이 매우 작기 때문에 **분모가 $0$에 매우 가까워집니다.**
2. 분모가 극단적으로 작아지면서 전체 값인 **그래디언트 $\frac{\nabla_\theta \pi_\theta(y_i)}{\pi_\theta(y_i)}$가 폭발적으로 커집니다.**


* **결과:** 실제 모델의 학습과 파라미터 변화는 오직 이 **Tail 토큰 구역에 집중**되어 일어납니다.


## 3.7 - 성능 향상의 진짜 원인 탐색 2 - 칭찬이 아니라 '벌점'이 동력이다

[비교 실험]
- Positive-only ($A_i > 0$만 적용): 성능 향상 거의 없음 
- Negative-only ($A_i < 0$만 적용): Standard OPD와 동일한 성능 달성  
- Fixed Negative ($A_i = -0.5$ 고정): 하위 20% 토큰에 똑같이 $-0.5$만 줘도 OPD와 대등  

[결론]: OPD의 본질은 교사의 지식 전달이 아니라 학생이 얼결에 뱉은 불안정한 Tail 토큰을 음의 어드밴티지로 깎아내는 억제(Suppression) 작업에 불과함.  

## 4-1. 제안 방법론


### 1. OPSA (On-Policy Self-Adaptation) 핵심 개념

OPSA는 외부 교사 모델(Teacher), Ground-Truth(실제 정답 데이터), 외부 검증기(Verifier)를 **완전히 제거**하고, 오직 **학생 모델 자신의 내부 확률 분포**만을 활용하여 스스로 성능을 끌어올리는 **Self-Improvement(자기 개선) 학습 알고리즘**입니다.

* **기존 OPD 방식:**
학생 모델이 토큰을 생성 $\rightarrow$ 초대형 외부 교사 모델이 매 토큰마다 평가 신호(Advantage $A_i$) 계산 $\rightarrow$ 노이즈가 섞인 피드백으로 학생 학습
* **OPSA 방식:**
외부 교사 호출 없이, 학생 모델이 문장을 생성할 때 계산되는 **자신의 토큰별 예측 확률 분포 및 엔트로피(Entropy)** 정보만으로 자체적인 어드밴티지 신호 $A_i^{\text{self}}$를 즉석에서 구성하여 학습


## 4-2. OPSA의 3단계 파이프라인

1. 1단계 (Sampling): 학생 정책 $\pi_s$로 문장 궤적 생성  
2. 2단계 (Filtering): $\log \pi_s$가 하위 20%에 속하는 Tail 토큰 집합 $T_{tail}$ 추출  
3. 3단계 (Advantage Assignment): 선택된 Tail 토큰에 엔트로피 비례 동적 음수 어드밴티지 부여

## 4-3. 엔트로피(Entropy)의 수학적 의미와 계산


### 1. 수식 구성요소 하나씩 뜯어보기

$$H_i = -\sum_{v \in V} \pi_s(v \mid x, y_{<i}) \log \pi_s(v \mid x, y_{<i})$$

* $H_i$: $i$번째 토큰 위치에서의 **엔트로피(불확실성) 값**
* $V$: 언어 모델이 가진 **전체 단어장(Vocabulary)** (예: 32,000개 ~ 128,000개의 토큰들)
* $v \in V$: 단어장에 존재하는 개별 단어/토큰 하나하나
* $x, y_{<i}$: 입력 프롬프트 $x$와 $i-1$번째까지 이미 생성된 문맥(Context)
* $\pi_s(v \mid x, y_{<i})$: 현재 문맥이 주어졌을 때, 학생 모델이 다음 단어로 **$v$라는 특정 단어를 선택할 확률 ($0 \le \pi \le 1$)**
* $\log \pi_s(...)$: 확률의 로그값 (확률이 낮을수록 $-\log \pi$는 매우 큰 양수가 됨 $\rightarrow$ '정보량' 또는 '놀람의 정도')

---

### 2. 왜 이 수식이 '불확실성의 크기'를 나타내는가?

#### ① 확률 분포의 상태에 따른 $H_i$ 값 비교

* **저엔트로피 ($H_i \to 0$, 확신하는 상태):**
* 모델이 특정 단어 하나(예: `15`)에 확률을 $99\%$ 몰아주고, 나머지 단어들의 확률은 $0%$에 가까운 상태입니다.
* $0.99 \times \log(0.99)$는 $0$에 가깝고, $0 \times \log(0)$ 역시 $0$으로 취급되므로 합산 결과 **엔트로피 $H_i$는 $0$에 매우 가깝습니다.**
* **의미:** "다음 단어는 무조건 이거다!" 하고 고민이 전혀 없는 상태.


* **고엔트로피 ($H_i \gg 0$, 헷갈리는 상태):**
* 모델이 단어 선택을 못 하고 여러 단어(예: `Therefore`, `However`, `Wait`)에 확률을 $25%$씩 똑같이 골고루 나누어 가진 상태입니다.
* 확률 분포가 평평(Uniform)할수록 $\sum -p \log p$의 계산 값은 **최댓값**을 향해 커집니다.
* **의미:** "여러 방향 중 어디로 가야 할지 모르겠다"며 선택지 사이에서 갈팡질팡하는 상태.



---

### 3. 언어 모델 추론에서의 실제 적용: 분기점 (Fork Tokens)

이 수식이 언어 모델의 풀이 과정(Reasoning Trajectory)에서 가지는 실제 의미는 다음과 같습니다.

#### ① 저엔트로피 구간 (단순 연산 / 무난한 문법)

* **예시:** `"2 + 3 = [ ]"` 뒤에 들어갈 토큰, 또는 `"I am a [ student ]"`
* 이미 앞 문맥에 의해 다음 나올 단어가 사실상 고정되어 있어 엔트로피가 극도로 낮습니다.
* 이 구간은 모델이 지식을 단순히 출력하는 영역이므로 학습 신호를 크게 줄 필요가 없습니다.

#### ② 고엔트로피 구간 (분기점, Fork Tokens)

* **예시:** 복잡한 수학 문제를 풀다가 중간에 `"Therefore..."`, `"Wait, let me check..."`, `"However..."` 같은 **추론 전환 토큰**을 만난 시점
* 이 지점에서는 생각의 방향이 "정답으로 가는 올바른 경로"와 "오답/환각으로 빠지는 잘못된 경로"로 갈라지게 됩니다.
* 따라서 모델 내부의 확률 분포가 팽팽하게 대립하며 **엔트로피 $H_i$가 수직 상승**합니다.

---

### 4. OPSA에서 이 수식이 갖는 결정적 역할

OPSA는 이 엔트로피 수식($H_i$)을 통해 "모델이 생성한 수많은 토큰 중 어디에 매질(벌점)을 가해야 하는가?"를 정확히 알아냅니다.

1. **타겟팅:** 엔트로피가 높은 토큰 위치($H_i \gg 0$)는 '추론의 갈림길'이자 잘못된 길(Tail)로 빠질 위험이 가장 높은 곳입니다.
2. **동적 벌점 계산:** OPSA는 이 위치에서 선택된 불안정한 Tail 토큰에 엔트로피 $H_i$ 값에 비례하는 음수 어드밴티지($A_i = -\alpha \cdot H_i$)를 때립니다.
3. **결과:** 모델이 가장 헷갈려하는 분기점($H_i$가 높은 곳)에서 얼떨결에 선택했던 잘못된 추론 가지(Tail)를 강하게 잘라내어(Suppress), 모델이 항상 정확한 논리적 경로만 선택하도록 유도합니다.


## 4-5. OPSA의 동적 어드밴티지 수식

---

### 1. 첫 번째 수식: Min-Max 정규화 (Min-Max Normalization)

$$\tilde{H}_i = \frac{H_i - H_{\min}}{H_{\max} - H_{\min}}$$

#### ① 변수 의미

* $H_i$: $i$번째 토큰 지점에서 실제로 계산된 **원래 엔트로피 값**
* $H_{\min}, H_{\max}$: 해당 문장/배치(Batch) 내에서 나타난 엔트로피의 **최솟값과 최댓값**
* $\tilde{H}_i$: $0$과 $1$ 사이로 스케일링된 **정규화 엔트로피 값 ($\tilde{H}_i \in [0, 1]$)**

#### ② 왜 정규화를 수행하는가?

* 모델, 문맥, 단어장 크기에 따라 엔트로피 절대값의 스케일($H_i$)이 제각각 달라집니다.
* 스케일이 제각각이면 학습 스텝마다 기울기(Gradient)의 크기가 요동치기 때문에, 엔트로피 상대값에 따라 **상대적인 불확실성의 크기를 $0$ (최저 불확실성) ~ $1$ (최고 불확실성) 표준 범위로 압축**해 안정적인 학습 환경을 만듭니다.

---

### 2. 두 번째 수식: 동적 어드밴티지 변환 (Dynamic Advantage Assignment)

$$A_i^{\text{dyn}} = -0.5 - 0.5 \cdot \tilde{H}_i$$

이 식은 $0 \le \tilde{H}_i \le 1$ 범위를 갖는 정규화 엔트로피를 실제 음수 벌점 값으로 선형 변환(Linear Mapping)하는 함수입니다.

#### ① 엔트로피 값에 따른 $A_i^{\text{dyn}}$ 변화

1. **엔트로피가 상대적으로 가장 낮을 때 ($\tilde{H}_i = 0$, 고민이 적은 지점):**

$$A_i^{\text{dyn}} = -0.5 - 0.5 \times (0) = \mathbf{-0.5}$$


* 해당 위치의 Tail 토큰에 상대적으로 약한 벌점($-0.5$)을 부여합니다.


2. **엔트로피가 상대적으로 가장 높을 때 ($\tilde{H}_i = 1$, 고민이 깊은 분기점):**

$$A_i^{\text{dyn}} = -0.5 - 0.5 \times (1) = \mathbf{-1.0}$$


* 해당 위치의 Tail 토큰에 가장 강력한 벌점($-1.0$)을 부여합니다.



#### ② 왜 하필 $[-1.0, -0.5]$ 하한/상한 범위를 가지는가?

* **모든 Tail 토큰에 기본 감점($-0.5$) 보장:**
확률 하위 20%에 속하는 Tail 토큰들은 어차피 불안정한 오답 가능성이 높습니다. 따라서 엔트로피가 낮더라도 **최소 $-0.5$의 일정한 음수 억제 신호**를 주어 확실히 누릅니다.
* **불확실한 분기점 추가 징벌 (가산 벌점 $-0.5$):**
엔트로피가 올라갈수록추가 벌점을 얹어 **최대 $-1.0$까지 벌점을 폭증**시킵니다.
* **폭발 방지 (Gradient Clipping 효과):**
벌점 $A_i$가 $-\infty$로 가거나 제어 불가능하게 커지는 것을 막고 최대 $-1.0$으로 캡(Cap)을 씌움으로써, 학습 과정에서 **그래디언트 폭발(Gradient Explosion)이 일어나는 현상을 수학적으로 완벽히 예방**합니다.


## 4-6. 메커니즘 - 확률 질량 재분배 (Probability Mass Redistribution)


### 1. 확률 질량(Probability Mass)의 기본 법칙

언어 모델이 다음 토큰을 선택할 때 출력하는 Softmax 확률의 총합은 수학적으로 반드시 $1.0$ (100%)이어야 합니다.

$$\sum_{v \in V} \pi_s(v) = 1.0$$

이 $1.0$이라는 전체 확률의 크기를 '확률 질량(Probability Mass)'이라고 부릅니다. 이 질량은 어딘가로 사라지는 것이 아니라, 전체 단어장(Vocabulary) 안에서 서로 나누어 가지게 됩니다.

---

### 2. 엔트로피별 차등 억제와 확률 이동 과정

OPSA가 지정한 두 영역(저엔트로피 vs 고엔트로피)에서 확률 질량이 어떻게 움직이는지 비교해 볼 수 있습니다.

#### ① 저엔트로피 위치 (고민이 적은 명확한 구간)

* **상황:** Top-1 단어의 확률이 이미 $90\%$ 이상이고, 나머지 단어들은 거의 $0%$에 가까운 상태입니다.
* **작동:** 약한 벌점($A_i^{\text{dyn}} = -0.5$)만 부여되며, Slide 14에서 살펴본 수학적 특성($\nabla \log \pi \to 0$)에 의해 미분값이 거의 $0$이 됩니다.
* **결과:** Top-1 토큰이 가진 기존의 높은 확신도와 확률 질량이 파괴되지 않고 **그대로 유지**됩니다.

#### ② 고엔트로피 분기점 (추론의 갈림길 & 헷갈리는 구간)

* **상황:** 논리적 갈림길에서 정답 근처의 상위 후보 단어들(Head)과 엉뚱한 헛소리/환각 후보 단어들(Tail)이 한데 뒤엉켜 확률을 나누어 가진 상태입니다.
* **작동:** 최고 수치의 벌점($A_i^{\text{dyn}} = -1.0$)이 부여되며, 확률이 낮았던 Tail 토큰 영역에서 **분모 효과(Denominator Effect)로 인해 거대한 음의 그래디언트가 폭발**합니다.
* **결과:** 이상한 꼬리(Tail) 토큰들의 확률이 수직 하락하며 강하게 차단(Suppress)됩니다.

---

### 3. 핵심: 확률 질량 재분배 (Redistribution)

Tail 토큰들이 가지던 확률이 수직 하락하여 거의 $0\%$로 바닥을 치면, Softmax의 합산 법칙($\sum \pi = 1.0$)을 맞추기 위해 **빼앗긴 확률 질량이 갈 곳을 찾아 이동**해야 합니다.

* **재분배 방향:** 약한 벌점을 받거나 억제 대상에서 제외된 상위 유력 후보 단어들(Head 토큰들)에게 그 확률이 골고루 분산되어 옮겨갑니다.
* **의의 (자발적 사고 탐색):**
외부 교사가 특정 단어 하나만 정답이라고 강요하는 방식이 아니라, "엉뚱한 길만 차단"해 둠으로써 상위권의 괜찮은 여러 논리 후보들 사이에서 모델이 스스로 최선의 추론 경로를 자발적으로 탐색(Self-Exploration)할 수 있는 공간을 열어주게 됩니다.



## 4-7. OPSA의 전체 Loss 함수


### 1. Loss 함수 수식 완전 해부

$$\mathcal{L}_{\text{OPSA}} = -\mathbb{E}_{y \sim \pi_s} \left[ \frac{1}{\vert{}T_{\text{tail}}\vert{}} \sum_{i \in T_{\text{tail}}} A_i^{\text{dyn}} \log \pi_s(y_i \mid x, y_{<i}) \right]$$

* **$\mathbb{E}_{y \sim \pi_s}$ (기대값 / On-Policy Sampling):**
외부 데이터를 쓰지 않고, 오직 현재 학생 모델 정책 $\pi_s$가 스스로 생성(Sampling)해낸 응답 문장 $y$에 대해서 계산한다는 의미입니다.
* **$\sum_{i \in T_{\text{tail}}}$ (Tail 토큰 한정 합산):**
생성된 전체 문장의 모든 토큰을 학습시키는 것이 아니라, **로그 확률이 하위 20%에 속하는 불확실한 Tail 토큰 집합 $T_{\text{tail}}$에만** 마스킹(Masking)을 뚫어 손실값을 더합니다.
* **$\frac{1}{\vert{}T_{\text{tail}}\vert{}}$ (정규화 / Normalization):**
문장마다 추출된 Tail 토큰의 개수($\vert{}T_{\text{tail}}\vert{}$)가 다를 수 있으므로, **Tail 토큰 개수로 나누어 평준화**함으로써 Gradient의 크기가 문장 길이에 따라 흔들리지 않게 잡아줍니다.
* **$A_i^{\text{dyn}}$ (동적 음수 어드밴티지):**
Slide 19에서 정의한 엔트로피 비례 벌점 값입니다. ($A_i^{\text{dyn}} \in [-1.0, -0.5]$)
* **$\log \pi_s(y_i \mid x, y_{<i})$ (로그 확률):**
$i$번째 위치에서 실제로 뽑힌 토큰 $y_i$의 출력 가능성(Log Probability)입니다.

---

### 2. 이 수식이 실제로 작동하는 연산 과정 (Gradient 역전파)

이 손실 함수를 모델 매개변수 $\theta$에 대해 미분(Gradient)하여 경사하강법으로 업데이트할 때 일어나는 원리는 다음과 같습니다.

1. **상위 80% (Head 토큰) 마스킹:**
$i \notin T_{\text{tail}}$ 인 토큰들은 수식의 $\sum$ 합산 대상에서 아예 마스킹(0으로 처리)되어 Gradient 계산에서 완전히 배제됩니다.
* **효과:** 불필요한 연산량을 엄청나게 줄이고, 이미 확실한 Head 지식이 흔들리는 것을 원천 차단합니다.


2. **하위 20% (Tail 토큰) 음수 업데이트:**
$A_i^{\text{dyn}}$가 항상 음수($-1.0 \sim -0.5$)이므로, 전체 Loss 앞에 붙은 마이너스($-$)와 곱해지면서 양의 손실(Positive Loss)이 형성됩니다.
3. **역전파 결과:**
경사하강법(Loss를 줄이는 방향)으로 학습하면, 해당 Tail 토큰 $y_i$의 확률 $\pi_s(y_i)$를 **강제로 끌어내리는(Suppress) 방향**으로 파라미터가 업데이트됩니다.

## 5-3. 반성적 장문 추론 능력의 발현

### 1. 주요 발현 현상 (Key Emergent Behaviors)

1. **'반성적 단어(Reflective Tokens)'의 비중 급증**
* 학습이 진행됨에 따라 모델은 답변 생성 도중 **`wait` (잠시만), `but` (하지만), `perhaps` (어쩌면), `let me check` (다시 확인해보자)** 같은 자가 검증 키워드를 생성하는 비율이 비약적으로 상승합니다.
* 이는 모델이 한 번에 직진으로 정답을 출력하려 하지 않고, **중간 단계에서 자신의 논리적 오류 가능성을 인지하고 되돌아보는 습관**을 스스로 형성했음을 보여줍니다.


2. **추론 길이(Thinking Length)의 자발적 확장**
* OPSA는 긴 답변을 쓰라고 강제하지 않았음에도, 모델의 평균 생성 토큰 길이가 크게 늘어납니다 (Long-form reasoning).
* 모델이 복잡한 문제를 만났을 때 가지치기(Pruning)를 반복하며 더 깊고 긴 사고 체인(Chain-of-Thought)을 전개하게 됩니다.



---

### 2. OPSA가 이 현상을 유도하는 작동 원리

이유는 OPSA의 핵심 메커니즘인 "불확실성이 높은 토큰의 억제"에 있습니다.

* **분기점(Forking Points)에서의 엔트로피 감지**:
모델이 실수나 잘못된 방향으로 빠지는 지점(Tail tokens)은 보통 **Log-probability가 낮고 엔트로피(Entropy)가 매우 높은 불안정한 분기점**입니다.
* **불확실한 경로의 억제 및 경로 재배치**:
OPSA는 이 고엔트로피 구간에 강한 음의 신호(Negative Advantage)를 주어, 모델이 섣불리 엉뚱한 결론을 내리거나 환각(Hallucination)에 빠지는 것을 막습니다.
* **반성적 탐색으로의 전환**:
잘못된 경로가 억제되면서 모델은 대체 경로를 찾게 되고, 이 과정에서 "이 길이 아니니 다시 검토하자(`wait`)"는 식의 반성적 토큰을 거쳐 올바른 논리적 경로로 수정해 나가는 패턴이 강하게 정착됩니다.

---

### 3. 결론적 의의

이 결과는 매우 중요한 시사점을 줍니다.
LLM에게 "생각해라", "스스로 검증해라"라는 직접적인 지침(Prompt)이나 정답 데이터(SFT/RLVR)를 주지 않더라도, **자신의 불확실한 오류 경로를 억제하는 적응형 강화학습(Self-Adaptation)만으로도 모델 내부에서 자발적으로 '자기 반성적 장문 추론(Reflective Long-Form Reasoning)' 알고리즘이 창발(Emergence)된다는 것**을 입증한 것입니다.


## 5-4. ORSA는 모델의 다양성을 훼손하지 않는다

### 1. 문제 의식: 기존 강화학습의 한계 (Mode Collapse)

* **기존 강화학습(RL/RLVR)의 문제**: 보통 모델을 온폴리시(On-policy) 강화학습으로 지도·단련하다 보면, 모델이 정답 확률이 높은 극히 일부의 '안전한 답변 경로'로만 쏠리는 **모드 붕괴(Mode Collapse)** 현상이 발생합니다.
* **결과**: 정확도는 올라갈지 몰라도, 답변의 표현 형태나 생성 경로의 **다양성(Diversity)이 급격히 저하**되는 부작용을 겪게 됩니다.

---

### 2. OPSA의 핵심 발견 (Key Finding)

* **다양성 유지/증가**: OPSA는 정답 라벨 없이 모델 자체의 고엔트로피(불확실한) 토큰을 선택적으로 억제하는 방식을 사용함에도 불구하고, **생상되는 답변 샘플 간의 다양성이 전혀 떨어지지 않거나 오히려 유지·향상**됨을 확인했습니다.
* **Pass@N 지표에서의 증명**:
* 문제당 $N$개의 응답을 생성할 때, 다양성이 높은 모델일수록 $N$이 커짐에 따라 정답을 맞힐 확률(Pass@N)이 상향 곡선을 그립니다.
* OPSA로 학습된 모델은 **Pass@1부터 Pass@32까지 지속적으로 가파른 성능 상승**을 보여, 모델이 매번 서로 다른 유효한 추론 경로(Reasoning Path)를 탐색할 능력을 유지하고 있음을 입증했습니다.



---

### 3. 다양성이 유지되는 이유

1. **'오답 경로'만 핀포인트 억제**:
* OPSA는 전체 답변을 무조건 깎아내리는 것이 아니라, 모델이 확신을 갖지 못하는 **하위 20%의 고엔트로피(불확실성) 토큰**에 대해서만 선택적으로 적응형 음의 보상(Adaptive Negative Advantage)을 부여합니다.


2. **다양한 '성공 경로'의 보존**:
* 올바른 탐색 방향이나 다양한 표현 양식 형태는 억제하지 않기 때문에, 모델이 문제를 해결하기 위해 **여러 가지 접근법(CoT)으로 시도할 수 있는 자유도**가 그대로 유지됩니다.



---

### 4. 한 줄 요약

> **"OPSA는 비효율적이거나 오류 가능성이 높은 억제 대상 토큰만 정밀하게 쳐내기 때문에, 기존 RL 기법들과 달리 모델의 독창적이고 다양한 추론 경로 탐색 능력을 고스란히 보존한다."**

## 5-5. 최적 학습 토큰 비율 

---

### 1. 실험 배경 및 목적

* OPSA는 답변 전체를 오답 처리(Negative Penalty)하는 대신, 불확실성이 높은(Low Log-Probability / High Entropy) "불안정한 핵심 토큰"만을 선택적으로 억제합니다.


* 따라서 '학습 대상 토큰의 비율(Fraction of Trained Tokens, $\rho$)'이 성능에 미치는 영향을 파악하기 위해 $\rho = 10\%, 20\%, 50\%, 100\%$ 등의 조건에서 비교 실험을 수행했습니다.



---

### 2. 주요 실험 결과 (Key Findings)

* **최적의 비율: $\rho = 20\%$**
* 전체 생성 토큰 중 **하위 20%의 토큰만 학습(억제)시켰을 때** 수학 추론 및 평가 벤치마크에서 가장 높고 안정적인 성능을 기록했습니다.




* **비율이 너무 낮을 때 ($\rho = 10\%$ 이하)**
* 모델이 추론 과정에서 저지르는 실수나 오류 분기점(Forking points)을 충분히 감지하고 억제하지 못해 성능 개선 폭이 미미했습니다.




* **비율이 너무 높을 때 ($\rho = 50\% \sim 100\%$)**
* **100% (모든 토큰 학습)**: 올바르게 추론하고 있는 정상적인 토큰까지 과도하게 음의 보상(Negative Advantage)을 받아, 모델의 유용한 표현 표현력과 안정성이 오염되는 현상(Degradation)이 발생했습니다.


* 학습 토큰 비율이 늘어날수록 성능 상승세가 둔화되거나 오히려 하락하는 경향을 보였습니다.





---

### 3. 메커니즘 해석 및 의의

1. **정밀 타격(Surgical Penalty)의 중요성**
* 추론 오류는 전체 답변 중 일부 불안정한 논리적 결함 지점(Tail Probability Tokens)에서 시작됩니다.


* 하위 20%만 핀포인트로 억제하는 것이 올바른 문맥 흐름은 보존하면서 오답으로 새는 길목만 효과적으로 차단하는 스윗 스폿(Sweet Spot)임을 입증했습니다.




2. **노이즈 제어와 연관성**
* 학습 대상 토큰 비율을 적절히 제한하는 것이 OPSA의 제목이기도 한 "Noisy Teacher(노이즈가 낀 자체 교사 신호)"로부터 오염을 방지하고 Self-Improvement(자가 발전)를 이루어내는 핵심 장치임을 보여줍니다.
